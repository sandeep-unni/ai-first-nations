```python
from deepforest import main
import torch
from PIL import Image
import numpy as np
import pandas as pd
import cv2
import matplotlib.pyplot as plt

%matplotlib inline
```

```python
# load deep forest model
model = main.deepforest()
model.load_model(model_name="weecology/deepforest-tree")
model.to(torch.device("cpu"))
```

```python
img = Image.open('./test_image.jpg').convert('RGB')
img = np.array(img)
```

```python
pred = model.predict_image(img)

pred.head()
```

```python
for index, row in pred.iterrows():
    cv2.rectangle(img, (int(row["xmin"]), int(row["ymin"])), (int(row["xmax"]), int(row["ymax"])), (0, 0, 0), thickness=10, lineType=cv2.LINE_AA)

plt.figure(figsize=(15,15))
cv2.imshow("deepforest output", np.flip(img,2))
plt.show()
```
